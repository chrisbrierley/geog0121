```python
# Import modules and packages
import xarray as xa
import numpy as np
import pandas as pd
```

```python
# Define file paths
UKESM_bc_file = '~/Documents/geog0121/data/UKESM1_model_output/additional_files/loadbc_Eday_UKESM1-0-LL_historical_r1i1p1f2_gn_19500101-20141230.nc'
UKESM_oa_file = '~/Documents/geog0121/data/UKESM1_model_output/additional_files/loadoa_Eday_UKESM1-0-LL_historical_r1i1p1f2_gn_19500101-20141230.nc'
# Open datasets
UKESM_bc = xa.open_dataset(UKESM_bc_file)
UKESM_oa = xa.open_dataset(UKESM_oa_file)
```

```python
# Have a look at data
print(UKESM_bc)
```

```python
# Take a monthly mean of the dataset and resample such that the output is listed at month start
UKESM_bc_mon = UKESM_bc.resample(time = 'MS').mean()
UKESM_oa_mon = UKESM_oa.resample(time = 'MS').mean()
# Have a look at data post-processing.
print(UKESM_bc_mon)
```

```python
# Set the metadata of the new dataset to be equal to the old dataset
UKESM_bc_mon.attrs = UKESM_bc.attrs
UKESM_oa_mon.attrs = UKESM_oa.attrs
# Set the metadata of the main variable to be the same as the old main variable
UKESM_bc_mon['loadbc'].attrs = UKESM_bc['loadbc'].attrs
UKESM_oa_mon['loadoa'].attrs = UKESM_oa['loadoa'].attrs
# Set the metadata changed as part of the mean calaculation to new values
UKESM_bc_mon.attrs['frequency'] = 'mon'
UKESM_bc_mon.attrs['table_id'] = 'Emon'
UKESM_oa_mon.attrs['frequency'] = 'mon'
UKESM_oa_mon.attrs['table_id'] = 'Emon'
```

```python
# Output new dataset as netcdf in current directory
UKESM_bc_mon.to_netcdf("loadbc_Emon_UKESM1-0-LL_historical_r1i1p1f2_gn_195001-201412.nc")
UKESM_oa_mon.to_netcdf("loadoa_Emon_UKESM1-0-LL_historical_r1i1p1f2_gn_195001-201412.nc")
```
